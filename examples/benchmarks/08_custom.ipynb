# Custom Checks Benchmark Results

Detectors for user-defined callable and remote endpoint checks.

```
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

GROUPS = ['custom']
df = pd.read_csv(Path('../../examples/benchmarks/results.csv'))
slugs = ['callable_check', 'remote_check']
df = df[df['detector_slug'].isin(slugs)].sort_values('f1', ascending=False)
display(df[['detector_slug','dataset','precision','recall','f1','wall_time_s']])
```

```
fig, ax = plt.subplots(figsize=(10, max(3, len(df)*0.35)))
colors = ['#7FB394' if f >= 0.8 else '#D9B566' if f >= 0.6 else '#E07B6E' for f in df['f1']]
ax.barh(df['detector_slug'], df['f1'], color=colors)
ax.axvline(0.8, color='#7FB394', linestyle='--', alpha=0.7, label='F1=0.8')
ax.axvline(0.6, color='#D9B566', linestyle='--', alpha=0.7, label='F1=0.6')
ax.set_title('Custom Checks — F1 Scores')
ax.set_xlabel('F1')
ax.set_xlim(0, 1.05)
ax.legend()
plt.tight_layout()
plt.show()
```
