# Basic Checks Benchmark Results

Aggregate-based checks for volume, completeness, freshness, range, and format.

```
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

GROUPS = ['basic']
df = pd.read_csv(Path('../../examples/benchmarks/results.csv'))
slugs = ['cardinality_in_range', 'column_pair_comparison', 'completeness', 'composite_uniqueness', 'date_format', 'date_part_missing_fraction', 'freshness_seconds_behind', 'max_in_range', 'median_in_range', 'min_in_range', 'monotonicity', 'null_fraction', 'numeric_mean', 'quantile_in_range', 'regex_match', 'row_count_in_range', 'set_exclusion', 'set_membership', 'sql_assertion_violation', 'stddev_in_range', 'string_case_violation', 'string_length_range', 'sum_in_range', 'uniqueness', 'validity', 'value_in_range', 'volume']
df = df[df['detector_slug'].isin(slugs)].sort_values('f1', ascending=False)
display(df[['detector_slug','dataset','precision','recall','f1','wall_time_s']])
```

```
fig, ax = plt.subplots(figsize=(10, max(3, len(df)*0.35)))
colors = ['#7FB394' if f >= 0.8 else '#D9B566' if f >= 0.6 else '#E07B6E' for f in df['f1']]
ax.barh(df['detector_slug'], df['f1'], color=colors)
ax.axvline(0.8, color='#7FB394', linestyle='--', alpha=0.7, label='F1=0.8')
ax.axvline(0.6, color='#D9B566', linestyle='--', alpha=0.7, label='F1=0.6')
ax.set_title('Basic Checks — F1 Scores')
ax.set_xlabel('F1')
ax.set_xlim(0, 1.05)
ax.legend()
plt.tight_layout()
plt.show()
```
